```python
df["Age"] =  df["Age"].fillna(df["Age"].median())
df.drop(columns=["Cabin"], inplace=True)
```

```python
df.dropna(subset=["Embarked"], inplace= True)
```

```python
print(df.isnull().sum())
```

```python
survive = df["Survived"].mean()
print(survive * 100)
```

```python
survive_rate = df.groupby("Sex")["Survived"].mean()
print(survive_rate * 100)
```

```python
srate_pclass = df.groupby("Pclass")["Survived"].mean()
print(srate_pclass *)
```

```python
sur_rate = df.groupby(["Pclass", "Sex"])["Survived"].mean()
print(sur_rate * 100)
```

```python
df["AgeGroup"] = pd.cut(df["Age"], bins=[0,10,18,35,60,100], labels=["child","Teen","Adult","MiddleAge","Senior"])
print((df.groupby("AgeGroup")["Survived"].mean()) * 100)
```

```python
df["Familysize"] = df["SibSp"] + df["Parch"]
family_srate = df.groupby("Familysize")["Survived"].mean()
print(family_srate *100)
```

```python
sur_pay = df.groupby("Survived")["Fare"].mean()
print(sur_pay)
```
